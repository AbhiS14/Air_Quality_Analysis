12 .Create:

pollution_level

Conditions:

High → Values > 300
Medium → 100 to 300
Low → below 100

```python
from pyspark.sql.functions import when
df.withColumn("Pollution_Level",
               when(df["Values"] > 300 ,"High")
               .when((df["Values"] >= 100) & (df["Values"] <=300),"Medium")
               .otherwise("Low") ) . show()
```

20. Create Environmental Monitoring Dashboard Dataset
Objective

Industry-level reporting.

Task

Generate location-wise report containing:

Average pollution
Maximum pollution
Total records

```python
from pyspark.sql.functions import avg,max,round,count
location_report=df.groupBy("Locations").agg(
    round(avg("Values"),2).alias("Average Pollution"),
    max("Values").alias("Maximum Pollution"),
    count("*").alias("Total Records")
).show()
```
